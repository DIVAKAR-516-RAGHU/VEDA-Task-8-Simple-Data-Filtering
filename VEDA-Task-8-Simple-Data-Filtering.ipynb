## Simple Data Filtering

### Objective
Create a collection of student records and filter students based on marks, age, and department.

### Tools
- Python
- Jupyter Notebook
- Pandas

# Student records

students = [
    {"name": "Divakar", "age": 20, "department": "ECE", "marks": 88},
    {"name": "Harini", "age": 21, "department": "CSE", "marks": 92},
    {"name": "Keerthana", "age": 20, "department": "ECE", "marks": 76},
    {"name": "Dharaneesh", "age": 22, "department": "EEE", "marks": 85},
    {"name": "Arun", "age": 19, "department": "CSE", "marks": 68},
    {"name": "Priya", "age": 21, "department": "ECE", "marks": 95}
]

# Display all student records
for student in students:
    print(student)

## 1. Filter Students Based on Marks

Find students who scored 85 or above.

```python
# Filter students with marks >= 85

high_scorers = []

for student in students:
    if student["marks"] >= 85:
        high_scorers.append(student)

print("Students with marks >= 85:")

for student in high_scorers:
    print(student)
```

## 2. Filter Students Based on Age

Find students who are 20 years old or younger.

```python
# Filter students with age <= 20

young_students = []

for student in students:
    if student["age"] <= 20:
        young_students.append(student)

print("Students aged 20 or below:")

for student in young_students:
    print(student)
```

## 3. Filter Students Based on Department

Find all students belonging to the ECE department.

# Filter ECE students

ece_students = []

for student in students:
    if student["department"] == "ECE":
        ece_students.append(student)

print("ECE Students:")

for student in ece_students:
    print(student)

## 4. Multiple Filtering Conditions

Find ECE students who scored 80 or above.

```python
# Filter ECE students with marks >= 80

filtered_students = []

for student in students:
    if student["department"] == "ECE" and student["marks"] >= 80:
        filtered_students.append(student)

print("ECE students with marks >= 80:")

for student in filtered_students:
    print(student)
```

## 5. Convert Student Records into a Pandas DataFrame

```python
import pandas as pd

# Convert list of dictionaries into DataFrame
df = pd.DataFrame(students)

# Display DataFrame
df
```

## 6. Pandas Filtering Based on Marks

Display students who scored 85 or above.

```python
df[df["marks"] >= 85]
```

## 7. Pandas Filtering Based on Department

Display students from the ECE department.

```python
df[df["department"] == "ECE"]
```

## 8. Pandas Filtering Using Multiple Conditions

Display ECE students who scored 80 or above.

```python
df[(df["department"] == "ECE") & (df["marks"] >= 80)]
```

## 9. Filtering Based on Age and Marks

Find students who are 20 years old or younger and scored 80 or above.

```python
df[(df["age"] <= 20) & (df["marks"] >= 80)]
```

## 10. Filtering Using OR Condition

Find students who belong to either ECE or CSE.

```python
df[
    (df["department"] == "ECE") |
    (df["department"] == "CSE")
]
```

## 11. Filtering and Sorting

Filter students who scored 80 or above and sort them by marks in descending order.

```python
filtered_sorted = df[df["marks"] >= 80].sort_values(
    "marks",
    ascending=False
)

filtered_sorted
```

## 12. Final Filtered Student Output

Display ECE students who scored at least 80 marks.

```python
final_result = df[
    (df["department"] == "ECE") &
    (df["marks"] >= 80)
]

print("Final filtered students:")

final_result
```

# Interview Questions

## 1. What is data filtering?

Data filtering is the process of selecting only the records that satisfy one or more specified conditions.

For example, we can filter students whose marks are greater than or equal to 85.

## 2. How would you filter a Pandas DataFrame?

A Pandas DataFrame can be filtered using Boolean conditions.

For example:

`df[df["marks"] >= 85]`

For multiple conditions, `&` is used for AND and `|` is used for OR.

Example:

`df[(df["department"] == "ECE") & (df["marks"] >= 80)]`

## 3. What is the difference between filtering and sorting?

Filtering selects rows that satisfy a particular condition.

Sorting changes the order of rows based on the values in one or more columns.

Example of filtering:

`df[df["marks"] >= 85]`

Example of sorting:

`df.sort_values("marks", ascending=False)`

Therefore:

- Filtering → selects specific records.
- Sorting → rearranges the order of records.
